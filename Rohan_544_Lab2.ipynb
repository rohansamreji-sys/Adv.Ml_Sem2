```python
#importing the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_regression

X_data, y_data = make_regression(
    n_samples=1000,
    n_features=12,
    n_informative=10,
    noise=15,
    random_state=42
)

feature_names = [
    "Age",
    "Income",
    "Experience",
    "Education",
    "CreditScore",
    "Savings",
    "Debt",
    "PropertyValue",
    "LoanAmount",
    "MonthlyExpenses",
    "WorkHours",
    "Investment"
]

X = pd.DataFrame(X_data, columns=feature_names)
y = pd.Series(y_data, name="Target")

df = X.copy()
df["Target"] = y

print("First 5 rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)


#exploring the data

print("Column Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nLast 5 rows:")
display(df.tail())


print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

display(df.describe())


#feature vs target visualization

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Age", y="Target")
plt.title("Age vs Target")
plt.xlabel("Age")
plt.ylabel("Target")
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Income", y="Target")
plt.title("Income vs Target")
plt.xlabel("Income")
plt.ylabel("Target")
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="PropertyValue", y="Target")
plt.title("Property Value vs Target")
plt.xlabel("Property Value")
plt.ylabel("Target")
plt.show()


correlation_matrix = df.corr()

plt.figure(figsize=(14, 10))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()

# SEPARATE FEATURES AND TARGET

X = df.drop("Target", axis=1)
y = df["Target"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

# TRAIN-TEST SPLIT


from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)

#feature scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled Training Data:")
print(X_train_scaled[:5])


# baseline linear regression
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(X_train_scaled, y_train)

linear_pred = linear_model.predict(X_test_scaled)

print("First 10 Linear Regression Predictions:")
print(linear_pred[:10])

#model evaluation function
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

def evaluate_model(y_true, y_pred):

    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)

    return mae, mse, rmse, r2


linear_mae, linear_mse, linear_rmse, linear_r2 = evaluate_model(
    y_test,
    linear_pred
)

print("Linear Regression Results")
print("MAE :", linear_mae)
print("MSE :", linear_mse)
print("RMSE:", linear_rmse)
print("R²  :", linear_r2)



#LASSO REGRESSION WITH GRID SEARCH
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV

lasso = Lasso(max_iter=10000)

alpha_values = {
    "alpha": [0.0001, 0.001, 0.01, 0.1, 1, 10, 100]
}

lasso_grid = GridSearchCV(
    estimator=lasso,
    param_grid=alpha_values,
    cv=5,
    scoring="neg_mean_squared_error"
)

lasso_grid.fit(X_train_scaled, y_train)

print("Best Lasso Alpha:")
print(lasso_grid.best_params_)


#final lasso model

best_lasso_alpha = lasso_grid.best_params_["alpha"]

final_lasso = Lasso(
    alpha=best_lasso_alpha,
    max_iter=10000
)

final_lasso.fit(X_train_scaled, y_train)

lasso_pred = final_lasso.predict(X_test_scaled)

lasso_mae, lasso_mse, lasso_rmse, lasso_r2 = evaluate_model(
    y_test,
    lasso_pred
)

print("Lasso Regression Results")
print("Best Alpha:", best_lasso_alpha)
print("MAE :", lasso_mae)
print("MSE :", lasso_mse)
print("RMSE:", lasso_rmse)
print("R²  :", lasso_r2)

# RIDGE REGRESSION WITH GRID SEARCH
from sklearn.linear_model import Ridge

ridge = Ridge()

ridge_grid = GridSearchCV(
    estimator=ridge,
    param_grid=alpha_values,
    cv=5,
    scoring="neg_mean_squared_error"
)

ridge_grid.fit(X_train_scaled, y_train)

print("Best Ridge Alpha:")
print(ridge_grid.best_params_)


#final ridge model

best_ridge_alpha = ridge_grid.best_params_["alpha"]

final_ridge = Ridge(
    alpha=best_ridge_alpha
)

final_ridge.fit(X_train_scaled, y_train)

ridge_pred = final_ridge.predict(X_test_scaled)

ridge_mae, ridge_mse, ridge_rmse, ridge_r2 = evaluate_model(
    y_test,
    ridge_pred
)

print("Ridge Regression Results")
print("Best Alpha:", best_ridge_alpha)
print("MAE :", ridge_mae)
print("MSE :", ridge_mse)
print("RMSE:", ridge_rmse)
print("R²  :", ridge_r2)


#model comparison

results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Lasso Regression",
        "Ridge Regression"
    ],
    "MAE": [
        linear_mae,
        lasso_mae,
        ridge_mae
    ],
    "MSE": [
        linear_mse,
        lasso_mse,
        ridge_mse
    ],
    "RMSE": [
        linear_rmse,
        lasso_rmse,
        ridge_rmse
    ],
    "R2": [
        linear_r2,
        lasso_r2,
        ridge_r2
    ]
})

display(results)


#coefficient comparison
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Linear Regression": linear_model.coef_,
    "Lasso Regression": final_lasso.coef_,
    "Ridge Regression": final_ridge.coef_
})

display(coefficients)


# zero coefficients in lasso

zero_lasso_features = coefficients[
    coefficients["Lasso Regression"] == 0
]

print("Features with zero Lasso coefficients:")
display(zero_lasso_features)


#lasso actual vs predicted
plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    lasso_pred
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.title("Lasso: Actual vs Predicted")
plt.xlabel("Actual Target")
plt.ylabel("Predicted Target")

plt.show()
#ridge actual vs predicted

plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    ridge_pred
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.title("Ridge: Actual vs Predicted")
plt.xlabel("Actual Target")
plt.ylabel("Predicted Target")

plt.show()


# coefficient visualization

plt.figure(figsize=(14, 7))

plt.plot(
    X.columns,
    linear_model.coef_,
    marker="o",
    label="Linear Regression"
)

plt.plot(
    X.columns,
    final_lasso.coef_,
    marker="o",
    label="Lasso"
)

plt.plot(
    X.columns,
    final_ridge.coef_,
    marker="o",
    label="Ridge"
)

plt.xlabel("Features")
plt.ylabel("Coefficient Value")
plt.title("Comparison of Regression Coefficients")

plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()

plt.show()
#LASSO CROSS-VALIDATION PERFORMANCE

lasso_cv_results = pd.DataFrame(
    lasso_grid.cv_results_
)

plt.figure(figsize=(8, 5))

plt.plot(
    lasso_cv_results["param_alpha"].astype(float),
    -lasso_cv_results["mean_test_score"],
    marker="o"
)

plt.xscale("log")
plt.xlabel("Alpha")
plt.ylabel("Mean CV MSE")
plt.title("Lasso Cross-Validation Performance")

plt.show()

#RIDGE CROSS-VALIDATION PERFORMANCE


ridge_cv_results = pd.DataFrame(
    ridge_grid.cv_results_
)

plt.figure(figsize=(8, 5))

plt.plot(
    ridge_cv_results["param_alpha"].astype(float),
    -ridge_cv_results["mean_test_score"],
    marker="o"
)

plt.xscale("log")
plt.xlabel("Alpha")
plt.ylabel("Mean CV MSE")
plt.title("Ridge Cross-Validation Performance")

plt.show()

#Final Summary


print("REGULARIZED REGRESSION FINAL SUMMARY")
print("\nDataset:")
print("Instances:", len(df))
print("Features :", X.shape[1])

print("\nBest Alpha Values:")
print("Lasso:", best_lasso_alpha)
print("Ridge:", best_ridge_alpha)

print("\nModel Performance:")
display(results)

print("\nNumber of features removed by Lasso:")
print(len(zero_lasso_features))

print("\nZero-coefficient Lasso features:")
print(zero_lasso_features["Feature"].tolist())
```


#INTERPRETATION OF REGULARIZATION


# Interpretation:
# Alpha controls the strength of regularization.
# A small alpha applies a weak penalty to the model coefficients.
# A large alpha applies a stronger penalty and shrinks coefficients more.

# Lasso uses L1 regularization and can reduce some coefficients exactly to zero. Therefore, Lasso can perform automatic feature selection.

# Ridge uses L2 regularization and generally keeps all features,but shrinks their coefficients toward zero.

# Feature scaling is important because regularization penalizes the size of coefficients, so features should be on a comparable scale.


# LASSO VS RIDGE SUITABILITY


# Interpretation:
# Lasso is useful when some features may be unnecessary because it can reduce their coefficients to exactly zero.

# Ridge is useful when most features may contribute to the target  and we want to keep them while controlling large coefficients.

# The final choice should be based on cross-validation performance,test-set performance, coefficient behaviour, and the project objective.

#  EXTENDED LASSO GRID SEARCH


# Self-learning:A wider alpha range is tested to check whether a better;regularization strength exists outside the original range.


```python
extended_alpha_values = {
    "alpha": [0.00001, 0.0001, 0.001, 0.01, 0.1, 1, 10, 100, 1000]
}

extended_lasso_grid = GridSearchCV(
    estimator=Lasso(max_iter=10000),
    param_grid=extended_alpha_values,
    cv=5,
    scoring="neg_mean_squared_error"
)

extended_lasso_grid.fit(X_train_scaled, y_train)

print("Extended Lasso Best Alpha:", extended_lasso_grid.best_params_["alpha"])
print("Extended Lasso Best CV MSE:", -extended_lasso_grid.best_score_)


# EXTENDED RIDGE GRID SEARCH


# Self-learning:
# The same wider alpha range is tested for Ridge so that both regularization methods are evaluated fairly.

extended_ridge_grid = GridSearchCV(
    estimator=Ridge(),
    param_grid=extended_alpha_values,
    cv=5,
    scoring="neg_mean_squared_error"
)

extended_ridge_grid.fit(X_train_scaled, y_train)

print("Extended Ridge Best Alpha:", extended_ridge_grid.best_params_["alpha"])
print("Extended Ridge Best CV MSE:", -extended_ridge_grid.best_score_)


# Compare the best cross-validation MSE obtained from the original and extended alpha ranges.

cv_comparison = pd.DataFrame({
    "Model": ["Lasso", "Extended Lasso", "Ridge", "Extended Ridge"],
    "Best CV MSE": [
        -lasso_grid.best_score_,
        -extended_lasso_grid.best_score_,
        -ridge_grid.best_score_,
        -extended_ridge_grid.best_score_
    ]
})

print(cv_comparison)

#FINAL EXTENDED LASSO AND RIDGE MODELS


# Train Lasso using the best alpha found from the extended grid search
final_extended_lasso = Lasso(
    alpha=extended_lasso_grid.best_params_["alpha"],
    max_iter=10000
)

final_extended_lasso.fit(X_train_scaled, y_train)

extended_lasso_pred = final_extended_lasso.predict(X_test_scaled)

extended_lasso_metrics = evaluate_model(
    y_test,
    extended_lasso_pred
)


# Train Ridge using the best alpha found from the extended grid search
final_extended_ridge = Ridge(
    alpha=extended_ridge_grid.best_params_["alpha"]
)

final_extended_ridge.fit(X_train_scaled, y_train)

extended_ridge_pred = final_extended_ridge.predict(X_test_scaled)

extended_ridge_metrics = evaluate_model(
    y_test,
    extended_ridge_pred
)

print("Extended Lasso Metrics:")
print("MAE:", extended_lasso_metrics[0])
print("MSE:", extended_lasso_metrics[1])
print("RMSE:", extended_lasso_metrics[2])
print("R2:", extended_lasso_metrics[3])

print("\nExtended Ridge Metrics:")
print("MAE:", extended_ridge_metrics[0])
print("MSE:", extended_ridge_metrics[1])
print("RMSE:", extended_ridge_metrics[2])
print("R2:", extended_ridge_metrics[3])



# Fix: Define linear_metrics, lasso_metrics, and ridge_metrics from previously calculated individual values
linear_metrics = (linear_mae, linear_mse, linear_rmse, linear_r2)
lasso_metrics = (lasso_mae, lasso_mse, lasso_rmse, lasso_r2)
ridge_metrics = (ridge_mae, ridge_mse, ridge_rmse, ridge_r2)


# fINAL MODEL COMPARISON


# Compare the baseline model with the regularized models.

final_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Lasso",
        "Ridge",
        "Extended Lasso",
        "Extended Ridge"
    ],
    "MAE": [
        linear_metrics[0],
        lasso_metrics[0],
        ridge_metrics[0],
        extended_lasso_metrics[0],
        extended_ridge_metrics[0]
    ],
    "MSE": [
        linear_metrics[1],
        lasso_metrics[1],
        ridge_metrics[1],
        extended_lasso_metrics[1],
        extended_ridge_metrics[1]
    ],
    "RMSE": [
        linear_metrics[2],
        lasso_metrics[2],
        ridge_metrics[2],
        extended_lasso_metrics[2],
        extended_ridge_metrics[2]
    ],
    "R2": [
        linear_metrics[3],
        lasso_metrics[3],
        ridge_metrics[3],
        extended_lasso_metrics[3],
        extended_ridge_metrics[3]
    ]
})

print(final_results)


#FINAL COEFFICIENT COMPARISON


# Compare the coefficients learned by the different models.

final_coefficients = pd.DataFrame({
    "Feature": feature_names,
    "Linear Regression": linear_model.coef_,
    "Lasso": final_lasso.coef_, # Fix: Changed from lasso_model.coef_
    "Ridge": final_ridge.coef_, # Fix: Changed from ridge_model.coef_
    "Extended Lasso": final_extended_lasso.coef_,
    "Extended Ridge": final_extended_ridge.coef_
})

print(final_coefficients)



#ZERO COEFFICIENTS IN EXTENDED LASSO


# Lasso can remove features by assigning their coefficients a value of zero.

zero_lasso_features = final_coefficients[
    final_coefficients["Extended Lasso"] == 0
]["Feature"]

print("Features with zero Extended Lasso coefficients:")
print(zero_lasso_features.tolist())

# Interpretation:
# Features with zero coefficients are not directly used by the final Lasso model for prediction.


#SELF-LEARNING: COMPARE CV PERFORMANCE
# The original and extended alpha ranges are compared.
# This helps determine whether expanding the search range can improve the cross-validation performance.

original_lasso_mse = -np.array(lasso_grid.cv_results_["mean_test_score"])
extended_lasso_mse = -np.array(
    extended_lasso_grid.cv_results_["mean_test_score"]
)

plt.figure(figsize=(10, 6))

plt.semilogx(
    lasso_grid.cv_results_["param_alpha"].data.astype(float),
    original_lasso_mse,
    marker="o",
    label="Original Lasso"
)

plt.semilogx(
    extended_lasso_grid.cv_results_["param_alpha"].data.astype(float),
    extended_lasso_mse,
    marker="o",
    label="Extended Lasso"
)

plt.xlabel("Alpha")
plt.ylabel("Cross-Validation MSE")
plt.title("Lasso Cross-Validation Performance")
plt.legend()
plt.grid(True)
plt.show()
```

# Interpretation:
# Lower cross-validation MSE indicates better validation performance.
# Grid search helps identify the alpha value that provides the best balance between model complexity and prediction error.

# Conclusion:
# Linear Regression provides the baseline performance.

# Lasso applies L1 regularization and can perform feature selection by reducing some coefficients to exactly zero.

# Ridge applies L2 regularization and reduces the magnitude of coefficients while generally retaining all features.

# GridSearchCV with 5-fold cross-validation is used to identify suitable alpha values for both Lasso and Ridge.

# The extended alpha experiment checks whether a wider search range can improve model performance.

# Scaling is performed before regularization because feature scales can affect the coefficient penalties.

# Model selection should consider MAE, MSE, RMSE, R2, cross-validation performance, and coefficient behaviour.
